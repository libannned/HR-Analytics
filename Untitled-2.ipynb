```python
# ======================================================================
# Hull Tactical Market Prediction – Full ML Pipeline (Corrected Version)
# Liban Dahir – COS 322 Project
# ======================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
)
from xgboost import XGBClassifier

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping


# ================================================================
# 1. LOAD DATA (train2.csv)
# ================================================================
train = pd.read_csv("/Users/najmamahad/Downloads/train2.csv")
test = pd.read_csv("/Users/najmamahad/Downloads/test.csv")


# ================================================================
# 2. CREATE TARGET
# market_forward_excess_returns = forward_returns - risk_free_rate
# target = 1 if market beats risk-free rate
# ================================================================
train["target"] = (train["market_forward_excess_returns"] > 0).astype(int)


# ================================================================
# 3. REMOVE LEAKAGE (future returns info)
# ================================================================
train = train.drop(columns=[
    "forward_returns",
    "risk_free_rate",
    "market_forward_excess_returns"
])


# ================================================================
# 4. HANDLE MISSING VALUES
# ================================================================
train = train.interpolate().ffill().bfill()


# ================================================================
# 5. SORT & TIME-BASED SPLIT
# ================================================================
train_sorted = train.sort_values("date_id")

y = train_sorted["target"]
X = train_sorted.drop(columns=["target"])

split_idx = int(len(train_sorted) * 0.8)

X_train = X.iloc[:split_idx]
X_test = X.iloc[split_idx:]
y_train = y.iloc[:split_idx]
y_test = y.iloc[split_idx:]


# ================================================================
# 6. SCALING
# ================================================================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


# ================================================================
# 7. PCA (retain 95% variance)
# ================================================================
pca = PCA(n_components=0.95)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)


# ================================================================
# 8. BASELINE MODELS
# ================================================================

# LOGISTIC REGRESSION
log_reg = LogisticRegression(max_iter=2000)
log_reg.fit(X_train_pca, y_train)
log_pred = log_reg.predict(X_test_pca)
log_prob = log_reg.predict_proba(X_test_pca)[:, 1]

# RANDOM FOREST
rf = RandomForestClassifier(n_estimators=300)
rf.fit(X_train_pca, y_train)
rf_pred = rf.predict(X_test_pca)
rf_prob = rf.predict_proba(X_test_pca)[:, 1]

# XGBOOST
xgb = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.9,
    colsample_bytree=0.9,
    objective="binary:logistic",
    eval_metric="logloss"
)
xgb.fit(X_train_pca, y_train)
xgb_pred = xgb.predict(X_test_pca)
xgb_prob = xgb.predict_proba(X_test_pca)[:, 1]


# ================================================================
# 9. AUTO-LSTM LOOKBACK WINDOW
# ================================================================
def create_sequences(data, target, window):
    X_seq, y_seq = [], []
    for i in range(window, len(data)):
        X_seq.append(data[i-window:i])
        y_seq.append(target[i])
    return np.array(X_seq), np.array(y_seq)


best_window = None
best_auc = -1

for window in [5, 10, 20, 30, 40]:
    X_seq, y_seq = create_sequences(X_train_scaled, y_train.values, window)
    X_seq_test, y_seq_test = create_sequences(X_test_scaled, y_test.values, window)

    model = Sequential([
        LSTM(64, input_shape=(window, X_train_scaled.shape[1])),
        Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer="adam", loss="binary_crossentropy")
    model.fit(X_seq, y_seq, epochs=3, batch_size=32, verbose=0)

    preds = model.predict(X_seq_test)
    auc = roc_auc_score(y_seq_test, preds)

    if auc > best_auc:
        best_auc = auc
        best_window = window

print("BEST LSTM WINDOW:", best_window)


# ================================================================
# 10. FINAL LSTM MODEL
# ================================================================
X_seq, y_seq = create_sequences(X_train_scaled, y_train.values, best_window)
X_seq_test, y_seq_test = create_sequences(X_test_scaled, y_test.values, best_window)

lstm_model = Sequential([
    LSTM(64, input_shape=(best_window, X_train_scaled.shape[1])),
    Dense(1, activation='sigmoid')
])
lstm_model.compile(optimizer="adam", loss="binary_crossentropy")

lstm_model.fit(
    X_seq, y_seq,
    epochs=12,
    batch_size=32,
    validation_split=0.2,
    callbacks=[EarlyStopping(patience=3, restore_best_weights=True)],
    verbose=1
)

lstm_prob = lstm_model.predict(X_seq_test).flatten()
lstm_pred = (lstm_prob > 0.5).astype(int)


# ================================================================
# 11. EVALUATION
# ================================================================
def evaluate(y_true, y_pred, y_prob):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1": f1_score(y_true, y_pred),
        "ROC-AUC": roc_auc_score(y_true, y_prob)
    }

results = {
    "Logistic Regression": evaluate(y_test, log_pred, log_prob),
    "Random Forest": evaluate(y_test, rf_pred, rf_prob),
    "XGBoost": evaluate(y_test, xgb_pred, xgb_prob),
    "LSTM": evaluate(y_seq_test, lstm_pred, lstm_prob),
}

print("\n=== FINAL MODEL RESULTS ===")
for model, metrics in results.items():
    print(model, metrics)


# ================================================================
# 12. BACKTEST
# ================================================================
raw_full = pd.read_csv("/Users/najmamahad/Downloads/train2.csv")

future_returns = raw_full["market_forward_excess_returns"].iloc[split_idx:].reset_index(drop=True)

xgb_strategy = np.where(xgb_pred == 1, future_returns, 0)
equity_curve = (1 + xgb_strategy).cumprod()

plt.figure(figsize=(10,5))
plt.plot(equity_curve, label="XGBoost Long-Only Strategy")
plt.title("Cumulative Returns Backtest")
plt.legend()
plt.grid(True)
plt.savefig("/Users/najmamahad/Downloads/backtest_equity_curve.png")
plt.show()


# ================================================================
# 13. SAVE RESULTS
# ================================================================
pd.DataFrame(results).to_csv("/Users/najmamahad/Downloads/model_results.csv")

print("\nSaved:")
print(" - /Users/najmamahad/Downloads/model_results.csv")
print(" - /Users/najmamahad/Downloads/backtest_equity_curve.png")
```

```python
# ======================================================================
# Hull Tactical Market Prediction – Improved Realistic Pipeline
# 5-Day Target • Lagged Features • Technical Indicators • No LSTM
# ======================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# =========================================================
# 1. LOAD DATA
# =========================================================
df = pd.read_csv("/Users/najmamahad/Downloads/train2.csv")

# =========================================================
# 2. CONSTRUCT 5-DAY TARGET (realistic)
# =========================================================
df["5d_forward_excess"] = df["market_forward_excess_returns"].shift(-5)
df["target"] = (df["5d_forward_excess"] > 0).astype(int)

# Drop any future-leaking columns AFTER computing target
df = df.drop(columns=["5d_forward_excess"])

# =========================================================
# 3. LAGGED FEATURES (no leakage)
# =========================================================
lag_features = ["forward_returns", "risk_free_rate", "market_forward_excess_returns"]

for col in lag_features:
    for l in [1, 2, 3, 5]:
        df[f"{col}_lag{l}"] = df[col].shift(l)

# =========================================================
# 4. TECHNICAL FEATURES (realistic)
# =========================================================
# Rolling mean & volatility
df["roll_mean_5"] = df["market_forward_excess_returns"].rolling(5).mean()
df["roll_vol_5"]  = df["market_forward_excess_returns"].rolling(5).std()

df["roll_mean_10"] = df["market_forward_excess_returns"].rolling(10).mean()
df["roll_vol_10"]  = df["market_forward_excess_returns"].rolling(10).std()

# Momentum features
df["momentum_5"] = df["market_forward_excess_returns"] - df["market_forward_excess_returns"].shift(5)
df["momentum_10"] = df["market_forward_excess_returns"] - df["market_forward_excess_returns"].shift(10)

# Drop rows with NaN from lagging/rolling
df = df.dropna().reset_index(drop=True)

# =========================================================
# 5. TRAIN / TEST SPLIT (TIME)
# =========================================================
split = int(len(df) * 0.8)

train = df.iloc[:split]
test  = df.iloc[split:]

y_train = train["target"]
y_test  = test["target"]

X_train = train.drop(columns=[
    "target",
    "forward_returns",
    "risk_free_rate",
    "market_forward_excess_returns"
])

X_test = test.drop(columns=[
    "target",
    "forward_returns",
    "risk_free_rate",
    "market_forward_excess_returns"
])

# =========================================================
# 6. SCALING + PCA
# =========================================================
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

pca = PCA(n_components=0.95)
X_train_p = pca.fit_transform(X_train_s)
X_test_p = pca.transform(X_test_s)

# =========================================================
# 7. MODELS: LR, RF, XGB
# =========================================================
def evaluate_model(model, X_train, y_train, X_test, y_test):
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    prob = model.predict_proba(X_test)[:,1]
    return {
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1": f1_score(y_test, pred),
        "ROC-AUC": roc_auc_score(y_test, prob)
    }

models = {
    "LR": LogisticRegression(max_iter=2000),
    "RF": RandomForestClassifier(
        n_estimators=300, 
        max_depth=8, 
        min_samples_leaf=5
    ),
    "XGB": XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss"
    )
}

results = {name: evaluate_model(m, X_train_p, y_train, X_test_p, y_test) for name, m in models.items()}

print("\n=== MODEL RESULTS (Realistic 5-Day Target Enhanced) ===")
print(results)

# =========================================================
# 8. REALISTIC BACKTEST (align signals to future)
# =========================================================
# Use XGB as main strategy
xgb = models["XGB"]
xgb.fit(X_train_p, y_train)
xgb_pred = xgb.predict(X_test_p)

# Properly aligned future returns
future_returns = test["forward_returns"].shift(-5).iloc[:-5]
signals = xgb_pred[:len(future_returns)]

# Strategy return
strategy_ret = np.where(signals == 1, future_returns, 0)
equity_curve = (1 + strategy_ret).cumprod()

plt.figure(figsize=(10,5))
plt.plot(equity_curve)
plt.title("Realistic XGBoost 5-Day Strategy Backtest")
plt.grid(True)
plt.savefig("/Users/najmamahad/Downloads/equity_curve_5day.png")
plt.show()
```
